## The perceptron

### Forward Propagation (순전파)

Input -> Weights -> Sum -> Non-Linearity -> Output

![Perceptron1](imagefile/perceptron1.png)

- Non-linear activation function (활성화 함수)는 생각보다 중요한 역할을 한다

- Single Neuron work에서 가장 중요한 것은 dot product of weights and inputs라고 한다.

![Perceptron2](imagefile/perceptron2.png)

- Bias는 입력과 곱해지지 않고 가중합에 더해진다.

- wx는 입력의 영향을 받지만, b는 기준점을 옮기는 역할을 한다 (shift 역할을 한다)

식으로 나타내자면

$$
\hat{y} = g\left(w_0 + \sum_{i=1}^{m} x_i w_i\right)
$$

$$
\hat{y} = g\left(w_0 + X^{T}W\right)
$$

where

$$
X=
\begin{bmatrix}
x_1\\
\vdots\\
x_m
\end{bmatrix},
\qquad
W=
\begin{bmatrix}
w_1\\
\vdots\\
w_m
\end{bmatrix}
$$

그렇기에 dot product이 중요하다고 말한다.


### Example

![percept](imagefile/percepex2.png)

Neural Network의 가장 기본 단위인 **Perceptron** 은 입력값에 가중치를 곱하고 합산한 뒤, 최종적으로 활성화 함수를 거쳐 출력값을 계산한다

입력 벡터와 가중치 벡터가 주어졌을 때, **Bias** (편향)를 포함한 가중치 합(Linear Combination)은 다음과 같이 행렬과 벡터의 곱으로 표현됩니다.

*   **Bias ($w_0$):** $1$

*   **Input Vector ($\boldsymbol{X}$):** $\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$

*   **Weight Vector ($\boldsymbol{W}$):** $\begin{bmatrix} 3 \\ -2 \end{bmatrix}$

가중치 합 연산을 적용한 최종 **Output** 수식은 다음과 같습니다.

$$ \hat{y} = g(w_0 + \boldsymbol{X}^T \boldsymbol{W}) = g\left(1 + \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}^T \begin{bmatrix} 3 \\ -2 \end{bmatrix}\right) = g(1 + 3x_1 - 2x_2) $$

여기서 $g$ 는 **Activation Function** 을 의미하며, $\hat{y}$ 은 최종 예측값이 된다

---

![percetp2](imagefile/percepex1.png)

### 기하학적 의미: Decision Boundary

**Activation Function** 내부에 있는 선형 식($1 + 3x_1 - 2x_2$)은 2차원 평면($x_1, x_2$) 공간에서 하나의 직선을 정의한다

### 핵심 개념
*   **Decision Boundary**: 가중치 합이 $0$ 이 되는 지점($1 + 3x_1 - 2x_2 = 0$)으로, 서로 다른 클래스를 구분하는 기하학적 기준선이 된다
*   **공간 분할 (Space Partitioning)**: 이 직선을 경계로 삼아 2차원 평면은 수학적 부호에 따라 두 개의 영역으로 양분된다
    *   양수 영역: $1 + 3x_1 - 2x_2 > 0$ 의 영역
    *   음수 영역: $1 + 3x_1 - 2x_2 < 0$ 의 영역

* 이제 이걸 활성화 함수에 넣어서 활용할 수 있다.


### 수업의 핵심: 단일 뉴런이 어떻게 작동하는가?

**dot product -> add bias -> apply a non-linearity**

## Non-Linear Activation Functions

신경망에서는 목적과 층의 위치에 따라 여러 가지 활성화 함수(Activation Function)를 사용한다.

![ActivateFunc](imagefile/ActivationFunc.png)

### 1. Sigmoid Function

$$

\sigma(x)=\frac{1}{1+e^{-x}}

$$

- 출력 범위: $(0,1)$

- 입력값을 확률처럼 표현할 수 있음

- 주로 이진 분류의 출력층에서 사용

- 입력의 절댓값이 크면 기울기가 0에 가까워지는 Gradient Vanishing 문제가 있음

#### Sigmoid의 도함수

$$
\sigma'(x)=\sigma(x)\left(1-\sigma(x)\right)
$$

도함수의 최댓값은 $x=0$에서 $1/4$이며, $|x|$가 커질수록 0에 가까워진다. 이것이 Gradient Vanishing이 발생하는 이유 중 하나다.

<br><br>

### 2. Hyperbolic Tangent

$$
\tanh(x)
=
\frac{e^x-e^{-x}}{e^x+e^{-x}}
$$

- 출력 범위: $(-1,1)$

- 출력의 중심이 0이므로 Sigmoid보다 학습에 유리할 수 있음

- 일부 RNN의 은닉 상태에서 사용

- Sigmoid와 마찬가지로 Gradient Vanishing 문제가 있음

#### tanh의 도함수

$$
\frac{d}{dx}\tanh(x)=1-\tanh^2(x)
$$

도함수의 최댓값은 $x=0$에서 1이며, $|x|$가 커지면 0에 가까워진다.

<br><br>


### 3. Rectified Linear Unit

$$
\operatorname{ReLU}(x)=\max(0,x)
$$

또는 조각함수로 표현하면

$$
\operatorname{ReLU}(x)
=
\begin{cases}
x, & x>0 \\
0, & x\leq0
\end{cases}
$$

- 출력 범위: $[0,\infty)$

- 계산이 간단하고 양수 영역에서 기울기가 1임

- Neural Network의 Hidden Layer에서 널리 사용

- 음수 영역의 기울기가 0이어서 뉴런이 학습하지 못하는 Dying ReLU 문제가 생길 수 있음

#### ReLU의 도함수

$$
\operatorname{ReLU}'(x)=
\begin{cases}
1, & x>0 \\
0, & x<0
\end{cases}
$$

$x=0$에서는 수학적으로 미분할 수 없다. 딥러닝 프레임워크에서는 구현상 보통 도함수를 0으로 정한다.

<br><br>


### 4. Why do we need activation functions?

- 비선형성을 부여하기 위해서 사용된다.

- 현실의 대부분의 데이터들은, 고차원적이고 비선형적이다.

- 복잡하고 뒤틀린 현실의 데이터를 근사할 수 있게 도움을 준다.

- 단순한 선형 경계를 넘어 조각별 선형 또는 비선형 결정 경계를 표현할 수 있음

<br>

![nonlinear1](imagefile/activate2.png)

현실의 데이터는 이보다도 더 고차원적이고 복잡하다

![nonlinear2](imagefile/activate1.png)

**The purpose of activation functions is to introduce non-linearities into the network**

- 선형함수만 사용하면 여러 층을 쌓아도 하나의 선형 변환으로 축약되므로 표현력이 증가하지 않는다.

- 비선형성을 도입하면 모델이 훨씬 복잡한 함수와 결정 경계를 근사할 수 있다.



## Multi Output Perceptron

![multithing](imagefile/multi1.png)

다중 출력 퍼셉트론은 여러 개의 출력 노드를 동시에 가지고 있어 다중 클래스 분류(Multi-class Classification)나 다차원 출력을 처리할 수 있다

---

### Network Architecture

그림의 예시는 $m$ 개의 입력값이 두 개의 출력 노드로 전달되는 구조를 보여준다

*   **Inputs**: $x_1, x_2, \dots, x_m$ 으로 총 $m$ 개의 입력 피처를 가진다

*   **Hidden Net Inputs**: 각 출력 노드로 들어가기 전, 가중치와 편향이 합산된 값인 $z_1$ 과 $z_2$ 가 계산된다

*   **Outputs**: 각 노드에서 **Activation Function** 인 $g$ 를 통과하여 최종 출력값 $y_1$ 과 $y_2$ 가 도출된다
    *   $y_1 = g(z_1)$
    *   $y_2 = g(z_2)$

---

### Mathematical Formulation

$$ z_i = w_{0,i} + \sum_{j=1}^{m} x_j w_{j,i} $$

---

### Summary

이 구조는 입력층과 출력층이 전결합(Fully Connected) 형태로 연결되어 있으며, 노드 개수만큼 독립적인 퍼셉트론 연산이 병렬로 수행되는 것과 같다

이 연산들이 모여 심층신경망(DNN)의 각 **Layer** 를 구성하는 핵심 기반이 된다


## TensorFlow와 PyTorch를 통한 구현

````python
import tensorflow as tf

class MyDenseLayer(tf.keras.layers.Layer): # tensorflow 상속
    def __init__(self, input_dim, output_dim):
        super(MyDenseLayer, self).__init__()

        # 학습 가능한 weight와 bias를 생성한다.
        self.W = self.add_weight(
            shape=(input_dim, output_dim), name="weight"
        )
        self.b = self.add_weight(
            shape=(output_dim,), initializer="zeros", name="bias"
        )
    
    def call(self, inputs):
        # Forward propagate the inputs
        z = tf.matmul(inputs, self.W) + self.b

        # Feed through a non-linear activation
        output = tf.math.sigmoid(z)
        return output
````

````python
import torch
import torch.nn as nn

class MyDenseLayer(nn.Module):
    def __init__(self, input_dim, output_dim):
        # Initialize weights and bias
        super(MyDenseLayer, self).__init__()
        # nn.Parameter는 기본적으로 requires_grad=True이다.
        self.W = nn.Parameter(torch.randn(input_dim, output_dim))
        self.b = nn.Parameter(torch.zeros(output_dim))

    def forward(self, inputs):
        # Forward propagate the inputs and apply the Sigmoid activation function
        z = torch.matmul(inputs, self.W) + self.b
        output = torch.sigmoid(z)
        return output

# 코드 설명
# 
````

## Pytorch 문법 정리



![torch1](imagefile/tfnn1.png)

````python
import torch
import torch.nn as nn

class MyDenseLayer(nn.Module):
    def __init__(self, input_dim, output_dim):
        super().__init__()
        # nn.Parameter와 크기 정의를 이 한 줄로 대체 (내부에 W, b가 이미 자동 포함됨)
        self.linear = nn.Linear(input_dim, output_dim)

    def forward(self, inputs):
        # 선형 연산 후 바로 시그모이드 활성화 함수 적용
        z = self.linear(inputs)
        output = torch.sigmoid(z)
        return output
````

## Single Layer Neural Network

![single](imagefile/singlehid.png)

> **Network flow**: Input ($m$) → Hidden Layer ($n$) → ReLU → Output (2)

입력에서 은닉층으로 이동할 때 $W_1$을, 은닉층에서 출력층으로 이동할 때 $W_2$를 사용한다. 중간의 ReLU가 비선형성을 추가하며, 최종 출력 두 개는 각 출력 노드의 점수다.

| 구간 | 연산 | 출력 shape | 학습 파라미터 |
|---|---|---|---|
| 입력 → 은닉층 | $Z_1 = XW_1 + b_1$ | `(batch_size, n)` | $W_1$, $b_1$ |
| 활성화 | $H = \operatorname{ReLU}(Z_1)$ | `(batch_size, n)` | 없음 |
| 은닉층 → 출력층 | $Y = HW_2 + b_2$ | `(batch_size, 2)` | $W_2$, $b_2$ |

### PyTorch 구현

```python
from torch import nn

m = 3  # 한 데이터가 가지는 입력 특징의 개수
n = 4  # 은닉층에 배치할 뉴런의 개수

model = nn.Sequential(
    nn.Linear(m, n),  # 입력 m개 → 은닉 뉴런 n개: XW₁ᵀ + b₁
    nn.ReLU(),        # 음수는 0으로 바꾸어 비선형성 추가
    nn.Linear(n, 2)   # 은닉 뉴런 n개 → 최종 출력 2개: HW₂ᵀ + b₂
)
```

### TensorFlow 구현

```python
import tensorflow as tf

m = 3  # 한 데이터가 가지는 입력 특징의 개수
n = 4  # 은닉층에 배치할 뉴런의 개수

model = tf.keras.Sequential([
    tf.keras.Input(shape=(m,)),                  # 입력 shape 명시
    tf.keras.layers.Dense(n, activation="relu"), # 입력 m개 → 은닉 뉴런 n개
    tf.keras.layers.Dense(2)                    # 은닉 뉴런 n개 → 최종 출력 2개
])

model.summary()  # 각 층의 출력 shape과 파라미터 개수 확인
```

> 두 구현은 같은 네트워크를 표현한다. 차이는 PyTorch의 `Linear`가 `(입력, 출력)`을 모두 받는 반면, Keras의 `Dense`는 출력 뉴런 수를 받고 입력 크기는 앞 레이어에서 자동으로 전달받는다는 점이다.


````python
import tensorflow as tf

m = 3  # 입력 특징의 개수
n = 4  # 은닉층 뉴런의 개수

model = tf.keras.Sequential([
    tf.keras.Input(shape=(m,)),
    tf.keras.layers.Dense(n, activation="relu"),
    tf.keras.layers.Dense(2)
])


model.summary()
````

````python
import torch
from torch import nn

m = 3 # 신경망의 입력 차원
n = 4 # 은닉층의 뉴런 수(hidden dimension)

model = nn.Sequential( # 여러 레이어를 순서대로 연결하여 하나의 신경망으로 만든다
    nn.Linear(m, n), # Linear Layer 생성
    nn.ReLU(), # activate function ReLU
    nn.Linear(n, 2) # 앞의 출력 차원의 것들을 입력 받아, 2개의 출력을 반환한다
)

x = torch.tensor([
    [1.0, 2.0, 3.0], # 각 행은 입력 특징 3개를 가진 하나의 데이터 샘플이다.
    [4.0, 5.0, 6.0]
])

output = model(x)

print(output) # 실제 예측값을 출력한다.
print(model) # 신경망의 전체적인 구조와 레이어 정보를 출력한다.
print(output.shape) # tensor의 형태(차원의 크기) 출력
````

## Deep Neural Network

![deepnn](imagefile/Deepnn.png)

기본 원리는 동일하다. 각 층의 출력 차원과 다음 층의 입력 차원을 맞추고, 은닉층 사이에 비선형 활성화 함수를 두면서 여러 층을 쌓는다.

<br>

### TensorFlow

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.Input(shape=(m,)),
    tf.keras.layers.Dense(n1, activation="relu"),
    tf.keras.layers.Dense(n2, activation="relu"),
    # 이런식으로 쭉 쌓아나가기
    tf.keras.layers.Dense(2)
])
```

<br>

### PyTorch

```python
from torch import nn

model = nn.Sequential(
    nn.Linear(m, n1),
    nn.ReLU(),
    nn.Linear(n1, n2),
    nn.ReLU(),
    # 같은 방식으로 입력·출력 차원을 맞추며 층을 쌓는다.
    nn.Linear(n2, 2)

)

```

````python
import torch
from torch import nn

# 1. 모델 구조 정의 (Deep Neural Network)
# 은닉층(Hidden Layer)을 여러 개 쌓아서 깊은(Deep) 신경망을 만든다
deep_model = nn.Sequential(
    nn.Linear(3, 16),  # 입력(3) -> 첫 번째 은닉층(16)
    nn.ReLU(),
    
    nn.Linear(16, 32), # 첫 번째 은닉층(16) -> 두 번째 은닉층(32)
    nn.ReLU(),
    
    nn.Linear(32, 8),  # 두 번째 은닉층(32) -> 세 번째 은닉층(8)
    nn.ReLU(),
    
    nn.Linear(8, 2)    # 세 번째 은닉층(8) -> 최종 출력(2)
)

# 2. 입력 데이터 생성 (Batch Size = 4)
# 4개의 데이터가 각각 3개의 특징(Feature)을 가지고 있습니다.
x_data = torch.tensor([
    [1.0, 2.0, 3.0],   # 1번 데이터
    [4.0, 5.0, 6.0],   # 2번 데이터
    [7.0, 8.0, 9.0],   # 3번 데이터
    [1.5, 2.5, 3.5]    # 4번 데이터
])

# 3. 순전파 (Forward Propagation) 수행
# 데이터를 모델에 통과시켜 예측값을 계산합니다.
output = deep_model(x_data)

# 4. 결과 확인
print("--- 입력 데이터의 형태 (Shape) ---")
print(x_data.shape)  # 4개의 데이터, 3개의 특징

print("\n--- 최종 출력 데이터 ---")
print(output)

print("\n--- 최종 출력 데이터의 형태 (Shape) ---")
print(output.shape)  # 4개의 데이터, 2개의 최종 예측값
````

![ex2](imagefile/ex2back.png)

신경망은 이 수업을 통과할 확률을 0.1, 즉 10%라고 예측한다

하지만 이 신경망의 예측은 틀렸다. 

신경망의 가중치는 무작위로 초기화된 상태로, 사실상의 아기같은 상태이다.

우리가 지금까지 한 것은 단순히 모델을 통과하는 forward pass(순전파)뿐이다.

아직 우리는 실제로 모델을 가르친 적이 없다.

핵심은 모델이 예측할 때마다, 예측에 대응하는 실제 정답이 훈련 데이터에 존재한다는 것이다.

모델이 예측한 0.1과 실제 정답 1 사이에는 차이가 존재한다.


![ex3](imagefile/ex3back.png)


- 신경망을 훈련하려면 모델에게 이 차이를 보여주어야 한다

- 모델이 언제 틀렸는지 알려줘야, 같은 입력이 있을 때 정답에 더 가까운 값을 출력한다.

- 모델이 같은 입력을 본다면 이전보다 개선된 답을 내놓아야 하며, 이 예시에서는 1에 더 가까운 값을 예측해야 한다.

예측값과 실제 값의 차이는 **loss(손실)** 이라는 값으로 정량화한다. 즉 이건 error이다.

손실이 작다는 것은 모델이 더 높은 품질의 답을 출력하고 있다는 뜻이다.

$$
\mathcal{L}\left(f(x^{(i)}; \boldsymbol{W}), y^{(i)}\right)
$$

위 식을 통해서 손실함수를 계산할 수 있다.



## Empirical Loss

![empiricalloss](imagefile/lossfun.png)

### 1. Concept

**Empirical Loss(경험적 손실)** 는 신경망이 전체 학습 데이터에 대해 얼마나 잘못 예측했는지를 하나의 값으로 나타낸 것이다.

각 데이터의 손실을 모두 계산한 뒤 평균을 구한다.

---

### 2. 전체 데이터셋

입력 데이터셋을 다음과 같이 나타낼 수 있다.

$$
X =
\begin{bmatrix}
4 & 5 \\
2 & 1 \\
5 & 8 \\
\vdots & \vdots
\end{bmatrix}
$$

각 행은 하나의 데이터 샘플이다.

예를 들어,

$$
x^{(1)} = [4, 5]
$$

$$
x^{(2)} = [2, 1]
$$

이다.

---

### 3. 신경망의 예측

신경망은 각 입력 데이터 $x^{(i)}$를 받아 예측값을 생성한다.

$$
\hat{y}^{(i)} = f(x^{(i)}; W)
$$

- $x^{(i)}$: $i$번째 입력 데이터

- $W$: 신경망의 가중치와 편향

- $f$: 신경망 모델

- $\hat{y}^{(i)}$: 신경망의 예측값

예측 결과는 다음과 같이 나타낼 수 있다.

$$
f(X) =
\begin{bmatrix}
0.1 \\
0.8 \\
0.6 \\
\vdots
\end{bmatrix}
$$

실제 정답은 다음과 같다.

$$
y =
\begin{bmatrix}
1 \\
0 \\
1 \\
\vdots
\end{bmatrix}
$$

---

### 4. 예측값과 실제값 비교

이진 분류에서는 일반적으로 예측값이 $0.5$ 이상이면 $1$, 미만이면 $0$으로 판단한다.

첫 번째 데이터

$$
\hat{y} = 0.1, \qquad y = 1
$$

예측 클래스는 $0$이고 실제 정답은 $1$이므로 틀린 예측이다.

두 번째 데이터

$$
\hat{y} = 0.8, \qquad y = 0
$$

예측 클래스는 $1$이고 실제 정답은 $0$이므로 틀린 예측이다.

세 번째 데이터

$$
\hat{y} = 0.6, \qquad y = 1
$$

예측 클래스와 실제 정답이 모두 $1$이므로 올바른 예측이다.

---

### 5. 개별 데이터의 Loss

각 데이터 하나에 대한 손실은 다음과 같이 계산한다.

$$
\mathcal{L}
\left(
f(x^{(i)}; W),
y^{(i)}
\right)
$$

즉,

$$
\mathcal{L}
\left(
\text{예측값},
\text{실제값}
\right)
$$

이다.

실제값이 $1$일 때 예측값이 $0.9$라면 손실이 작다.

$$
y = 1, \qquad \hat{y} = 0.9
$$

반대로 실제값이 $1$인데 예측값이 $0.1$이라면 손실이 크다.

$$
y = 1, \qquad \hat{y} = 0.1
$$

---

### 6. Empirical Loss 공식

전체 데이터에 대한 경험적 손실은 다음과 같다.

$$
J(W)
=
\frac{1}{n}
\sum_{i=1}^{n}
\mathcal{L}
\left(
f(x^{(i)}; W),
y^{(i)}
\right)
$$

각 데이터의 손실을 다음과 같이 정의하면,

$$
L_1 =
\mathcal{L}
\left(
f(x^{(1)}; W),
y^{(1)}
\right)
$$

$$
L_2 =
\mathcal{L}
\left(
f(x^{(2)}; W),
y^{(2)}
\right)
$$

$$
\vdots
$$

$$
L_n =
\mathcal{L}
\left(
f(x^{(n)}; W),
y^{(n)}
\right)
$$

전체 평균 손실은 다음과 같다.

$$
J(W)
=
\frac{L_1 + L_2 + \cdots + L_n}{n}
$$

---

### 7. 학습의 목표

신경망의 예측은 가중치 $W$에 따라 달라진다.

$$
W
\rightarrow
f(x; W)
\rightarrow
\hat{y}
\rightarrow
J(W)
$$

따라서 신경망 학습의 목표는 경험적 손실 $J(W)$를 최소로 만드는 가중치를 찾는 것이다.

$$
W^*
=
\operatorname*{arg\,min}_{W}
J(W)
$$

Gradient Descent에서는 가중치를 다음과 같이 갱신한다.

$$
W
\leftarrow
W
-
\eta
\nabla_W J(W)
$$

- $\eta$: learning rate
- $\nabla_W J(W)$: 가중치에 대한 손실 함수의 gradient

---

### 8. Loss Function과 Cost Function

두 개념은 비슷하지만 일반적으로 다음과 같이 구분한다.

### Loss Function

하나의 데이터 샘플에 대한 손실이다.

$$
\mathcal{L}
\left(
\hat{y}^{(i)},
y^{(i)}
\right)
$$

### Cost Function 또는 Objective Function

전체 데이터의 평균 손실이다.

$$
J(W)
=
\frac{1}{n}
\sum_{i=1}^{n}
\mathcal{L}
\left(
\hat{y}^{(i)},
y^{(i)}
\right)
$$

$J(W)$는 다음과 같은 이름으로도 불린다.

- Objective Function
- Cost Function
- Empirical Risk

---

### 9. 왜 Empirical이라고 부르는가?

우리가 궁극적으로 알고 싶은 것은 가능한 모든 데이터에 대한 평균 손실이다.

$$
\mathbb{E}_{(x,y)\sim P}
\left[
\mathcal{L}
\left(
f(x; W),
y
\right)
\right]
$$

하지만 실제 데이터 분포 $P$ 전체를 알 수는 없다.

따라서 실제로 수집한 학습 데이터만을 이용하여 평균 손실을 계산한다.

$$
\frac{1}{n}
\sum_{i=1}^{n}
\mathcal{L}
\left(
f(x^{(i)}; W),
y^{(i)}
\right)
$$

관측된 데이터, 즉 경험적으로 수집된 데이터를 사용하기 때문에 **Empirical Loss**라고 부른다.

---

### 10. 전체 과정

각 데이터에 대해 다음 과정이 진행된다.

$$
x^{(i)}
\rightarrow
f(x^{(i)}; W)
\rightarrow
\hat{y}^{(i)}
\rightarrow
\mathcal{L}
\left(
\hat{y}^{(i)},
y^{(i)}
\right)
$$

모든 데이터의 손실을 평균 내면 다음과 같다.

$$
J(W)
=
\text{전체 데이터 손실의 평균}
$$

신경망 학습은 결국 다음 과정이다.

$$
\boxed{
\text{가중치 } W \text{를 조정하여 } J(W) \text{를 최소화한다}
}
$$

## Binary Cross Entropy Loss (이진 크로스 엔트로피 손실)

![bcel](imagefile/BCEL.png)

### 1. 개요 (Overview)
* **목적:** 정답이 $0$ 또는 $1$인 **이진 분류(Binary Classification)** 문제에서 모델의 예측 확률과 실제 정답 사이의 오차를 계산한다

* 모델이 정답을 맞췄을 때는 손실(Loss)을 **0**에 가깝게 준다.
* 틀렸을 때는 틀린 정도에 따라 손실을 **무한대($\infty$)** 에 가깝게 큰 패널티를 부여하는 방식입니다.

---

### 2. 수학적 수식 분해
전체 데이터 세트에 대한 손실 함수 공식은 다음과 같다. 여기서 $\log$는 자연로그 $\ln$을 의미한다.

$$J(W) = -\frac{1}{n}\sum_{i=1}^{n} \left[ y^{(i)}\log(f(x^{(i)}; \boldsymbol{W})) + (1-y^{(i)})\log(1-f(x^{(i)}; \boldsymbol{W})) \right]$$


* $y$ : 실제 정답 레이블 ($0$ 또는 $1$)

* $\hat{y}$ 또는 $f(x)$ : 모델이 출력한 예측 확률 ($0$과 $1$ 사이의 소수)

전체 평균을 내는 기호($-\frac{1}{n}\sum$)를 제외하고, **데이터 1개에 대한 손실 공식** 만 떼어놓고 보면 정답($y$)의 값에 따라 식이 둘로 쪼개진다

① 실제 정답($y$)이 1인 경우
정답 $y=1$을 대입하면 우측 덩어리인 $(1-y)\log(1-\hat{y})$는 $(1-1)=0$이 되어 사라진다

* **남는 식:** $-\log(\hat{y})$

* 모델이 정답일 확률을 $1.0$(100%)이라고 잘 맞추면 $-\log(1.0) = 0$이 되어 오차가 없다. 

* 반대로 $0.0$에 가깝게 잘못 예측하면 오차가 폭발적으로 늘어납니다.

② 실제 정답($y$)이 0인 경우

정답 $y=0$을 대입하면 좌측 덩어리인 $y\log(\hat{y})$가 $0$이 되어 사라진다.

* **남는 식:** $-\log(1-\hat{y})$

* **해석:** 모델이 정답일 확률을 $0.0$(0%)이라고 잘 맞추면 $-\log(1-0) = 0$이 되어 오차가 없습니다. 반대로 $1.0$에 가깝게 잘못 예측하면 오차가 폭발적으로 늘어납니다.

---

### 3. 직관적인 연산 예시

예시 A: 크게 틀린 경우 (1행 데이터)

* **모델의 예측 ($\hat{y}$):** `0.1` (정답이 1일 확률이 10%라고 예측)

* **실제 정답 ($y$):** `1`

* **손실 계산:** $y=1$이므로 $-\log(\hat{y})$ 공식 적용

  $$\text{Loss} = -\ln(0.1) \approx \mathbf{2.3026}$$

* **결과:** 정답과 반대 방향으로 강하게 예측했으므로 **큰 패널티(약 2.3026)**가 부여됩니다. BCE와 딥러닝 프레임워크는 일반적으로 자연로그를 사용합니다. (이미지에서 $\times$ 표시된 부분)

예시 B : 비교적 잘 맞춘 경우 (3행 데이터)

* **모델의 예측 ($\hat{y}$):** `0.6` (정답이 1일 확률이 60%라고 예측)

* **실제 정답 ($y$):** `1`

* **손실 계산:** 동일하게 $-\log(\hat{y})$ 공식 적용

  $$\text{Loss} = -\ln(0.6) \approx \mathbf{0.5108}$$

* **결과:** 완벽하진 않지만 정답(1)에 더 가까운 방향으로 예측했으므로 **비교적 적은 패널티(약 0.5108)**가 부여됩니다. (이미지에서 $\checkmark$ 표시된 부분)

---

### 4. PyTorch 구현 — 출력이 하나인 이진 분류
```python
import torch.nn.functional as F

# 모델이 Sigmoid를 통과하지 않은 logit을 출력할 때 사용한다.
# 내부에 Sigmoid와 BCE가 함께 들어 있어 수치적으로 더 안정적이다.
loss = F.binary_cross_entropy_with_logits(predicted_logits, y.float())

```

### 5. TensorFlow 구현 — 출력이 하나인 이진 분류
```python
# predicted_logits는 Sigmoid를 통과하기 전 모델의 출력이다.
loss = tf.reduce_mean(
    tf.nn.sigmoid_cross_entropy_with_logits(
        labels=y,
        logits=predicted_logits
    )
)
```


## Mean Squared Error Loss (평균 제곱 오차 손실)

![msel](imagefile/MSEL.png)


### 1. 개요 (Overview)

* **목적:** 출력값이 0 또는 1의 확률이 아니라, 기온, 주가, 시험 점수처럼 **연속적인 실수(Continuous real numbers)** 를 예측하는 **회귀(Regression)** 문제에서 오차를 계산한다

* **핵심 원리:** 모델의 예측값과 실제 정답의 거리를 잰 뒤, 그 차이를 **제곱(Square)** 하여 평균을 낸다

* 오차가 클수록 손실이 오차의 제곱에 비례해 이차적으로 증가하므로 큰 오차에 더 큰 패널티를 준다

---

### 2. 수식
슬라이드 중심에 있는 MSE 공식은 다음과 같다

$$J(W) = \frac{1}{n}\sum_{i=1}^{n} \left( y^{(i)} - f(x^{(i)}; \boldsymbol{W}) \right)^2$$

* $y$ : 실제 정답 (Actual / Ground Truth)

* $f(x)$ 또는 $\hat{y}$ : 모델이 출력한 예측값 (Predicted)

* $n$ : 전체 데이터의 개수

### 공식의 작동 단계 (오른쪽 데이터 행렬 대입)

1. **차이 구하기 ($y - f(x)$):** 실제 정답과 모델 예측값의 순수한 빼기 오차를 구한다.

   * **row1 데이터:** 정답은 `90`인데 모델은 `30`으로 예측했습니다. 차이는 $90 - 30 = 60$ 입니다. (많이 틀림 $\times$)

   * **row3 데이터:** 정답은 `95`인데 모델은 `85`로 예측했습니다. 차이는 $95 - 85 = 10$ 입니다. (비교적 잘 맞춤 $\checkmark$)

<br>

2. **제곱하기 ($(\cdot)^2$):** 차이 값을 제곱한다 

   * 오차가 음수가 나오든 양수가 나오든 상관없이 전부 양수 데이터로 만든다

   * 1행의 오차는 $60^2 = 3600$이 되고, 3행의 오차는 $10^2 = 100$이 된다. 오차가 클수록 패널티가 어마어마하게 커진다

3. **평균 내기 ($\frac{1}{n}\sum$):** 모든 데이터에서 나온 제곱 오차들을 다 더한 뒤 데이터 개수($n$)로 나누어 평균값(단 하나의 숫자)을 낸다

---

### 3. 이진 크로스 엔트로피(BCE) vs 평균제곱오차(MSE) 직관적 비교

강의 내용과 연결하여 정리하면 다음과 같은 명확한 차이가 있습니다.

| 비교 항목 | Binary Cross Entropy (BCE) | Mean Squared Error (MSE) |
| :--- | :--- | :--- |
| **문제 유형** | **분류 (Classification)** (예/아니오) | **회귀 (Regression)** (연속적인 숫자) |
| **최종 출력층** | `0.0` ~ `1.0` 사이의 확률 값 | 제한 없는 연속적인 실수 값 |
| **채점 방식** | $\log$ 함수를 이용한 확률 분포 비교 | 실제 값의 차이를 제곱하여 거리 비교 |
| **대표적인 출력층** | 출력 1개인 logit 또는 확률 | 단일 회귀는 출력 1개, 다중 출력 회귀는 여러 개 |


---

### 4. PyTorch

```python
import torch

# 파이토치의 대표적인 회귀용 MSE 손실 함수
loss = torch.nn.functional.mse_loss(predicted, y)

```

### 5. TensorFlow

```python
loss = tf.reduce_mean(tf.square(tf.subtract(y, predicted)))
loss = tf.keras.losses.MeanSquaredError()(y, predicted)
```

## Loss Optimization (손실 최적화)
![train1](imagefile/train2.png)

### 1. 개요 (Overview)
* **목적:** 딥러닝 모델 학습의 궁극적인 목표를 나타낸다

* **핵심 원리:** 무작위 숫자로 시작했던 Weights들을 수정하며, 전체 데이터셋에서 발생하는 **오차의 평균(Loss)을 가장 낮게 만드는 최적의 가중치 조합** 을 찾아내는 과정

* 즉 J(W)를 최소화하는 가중치들을 찾고 싶어한다. 

---

### 2. 수식

슬라이드에 등장하는 수식은 뉴럴 네트워크가 학습을 통해 달성하고자 하는 최종 목적지이다

$$\boldsymbol{W}^* = \arg\min_{\boldsymbol{W}} \frac{1}{n}\sum_{i=1}^{n} \mathcal{L}(f(x^{(i)}; \boldsymbol{W}), y^{(i)})$$
$$\boldsymbol{W}^* = \arg\min_{\boldsymbol{W}} J(\boldsymbol{W})$$

각 기호가 의미하는 바를 뜯어보면 아주 직관적임

* $\boldsymbol{W}$ : 편향을 포함하여 뉴럴 네트워크가 가진 **모든 학습 가능 파라미터의 집합**이다


* $\boldsymbol{W}^*$ (W 스타) : 우리가 최종적으로 얻고 싶은 **'최적의 가중치(정답 가중치)'** 를 의미한다


* $\arg\min_{\boldsymbol{W}}$ : 뒤에 나오는 값을 **최소(minimum)로 만드는 입력 파라미터 $\boldsymbol{W}$를 찾으라**는 수학적 명령어


* $\mathcal{L}(f(x^{(i)}; \boldsymbol{W}), y^{(i)})$ : 데이터 1개에 대한 오차 함수. 모델의 예측값($f(x)$)과 실제 정답($y$)을 비교하여 채점한다 (앞서 배운 BCE나 MSE 함수가 이 자리에 들어감)


* $\frac{1}{n}\sum_{i=1}^{n}$ : 전체 데이터 $n$개에서 발생한 오차를 다 더해서 평균을 내겠다는 뜻이다


* $J(\boldsymbol{W})$ : 이 전체 데이터의 **경험적 평균 손실(Empirical average loss)** 덩어리를 하나의 함수 기호로 축약한 것이다

![imagewith](imagefile/train3.png)

- 파라미터가 단 두 개 뿐인 뉴럴 네트워크의 시각화 자료

- 두 파라미터가 어떤 조합을 갖느냐에 따라, 그래프의 높이에 해당되는 Loss값이 결정된다

- 우리의 목표는 Loss가 가장 낮은, w0과 w1이라는 두 가중치를 가진 뉴럴 네트워크를 찾아내는 것

이를 Neural Network의 Training이라고 한다


---

### 3. 강의 내용과의 연결 (직관적 이해)

1. **남는 단 하나의 숫자 ($J(\boldsymbol{W})$):** 수많은 데이터의 오차를 평균 낸 손실(Loss) 값은 결국 **딱 하나의 숫자** 로 귀결되며, 이 숫자는 가중치 $\boldsymbol{W}$가 변함에 따라 높낮이가 바뀌는 형태를 가진다

2. **산의 가장 낮은 바닥 찾기:** 파라미터 조합에 따라 오차가 높고 낮은 3차원 곡선 지도가 그려졌을 때, $\arg\min$은 **"가장 높이가 낮은 바닥 지점의 위도와 경도($w_0, w_1$)"**를 알아내라는 것과 같다

3. **경사하강법(Gradient Descent)의 출발점:** 이 최적의 지점 $\boldsymbol{W}^*$를 향해 가기 위해서, 우리는 가중치를 무작위로 초기화한 뒤 기울기의 반대 방향으로 한 걸음씩 내려가는 학습(Training) 과정을 시작하게 된다


## How to Train Neural Network

![train4](imagefile/train4.png)

### 1. Randomly initialize ($w_0, w_1$)

- $w_0, w_1$의 값을 랜덤하게 선택해서, 이 부분의 Loss를 구한다

### 2. Compute gradient

- Loss를 구한 이후 그 부분에서의 gradient(기울기)를 구한다.

- gradient는 바로 그 위치에서 경사면이 어떤 방향을 향하고 있는지 알려준다

- 국소적인(Local) 위치에서 어느 쪽이 올라가는 방향인지를 알려준다

### 3. Take small step

- 우리는 당연히 Loss를 줄이고 싶기에, 기울기에 마이너스를 곱해 반대 방향으로 small step

- 새 지점에서 다시 새로운 기울기를 계산하고, 기울기가 가리키는 방향의 음의 방향으로 다시 작은 한 걸음을 내딛는걸 반복한다

- 적절한 learning rate와 조건이 주어지면 이 과정을 반복하면서 손실이 더 작은 지점으로 수렴할 수 있다.

![train5](imagefile/train5.png)

다만 이 방법으로 계속 해도, 이는 전역에서의 최솟값(Global Minimum)이 아닐 수도 있다.

- 어디에서 시작하느냐에 따라 달라지기 때문이다.


## Gradient Descent(경사하강법)

### Algorithm

1. Initialize weights randomly $\sim \mathcal{N}(0, \sigma^2)$

2. Loop until convergence:

3. Compute gradient, $\frac{\partial J(\boldsymbol{W})}{\partial \boldsymbol{W}}$

4. Update weights, $\boldsymbol{W} \leftarrow \boldsymbol{W} - \eta \frac{\partial J(\boldsymbol{W})}{\partial \boldsymbol{W}}$

5. Return weights

### TensorFlow Code

```python
import tensorflow as tf

# 예제를 완전히 실행할 수 있도록 J(w)=(w-3)^2를 손실 함수로 사용한다.
# 이 함수는 w=3에서 최솟값 0을 가진다.
def compute_loss(weights):
    return tf.square(weights - 3.0)

# shape=[1]인 학습 가능한 가중치를 정규분포에서 초기화한다.
weights = tf.Variable(tf.random.normal(shape=[1]))
lr = 0.1

# 무한 루프 대신 정해진 횟수만 학습한다.
for step in range(20):
    # [2] tf.GradientTape()는 내부에서 일어나는 모든 연산을 테이프에 기록하는 장치이다
    #     이 컨텍스트 안에서 계산되는 연산들에 대해 자동으로 미분(Gradient)을 계산할 준비를 한다
    with tf.GradientTape() as g:
        # [3] 현재 가중치를 기반으로 비용 함수(Loss Function)인 compute_loss를 호출하여 손실 값을 계산한다
        loss = compute_loss(weights)
        
        # [4] 테이프(g)에 기록된 연산 과정을 바탕으로 loss에 대한 weights의 편미분 값(Gradient)을 계산한다
        #     수학적으로는 변수가 여러 개일 때의 다변수 미분(Gradient) 연산을 수행한다
        gradient = g.gradient(loss, weights)
    
    # [5] Variable을 유지하면서 W <- W - lr * Gradient를 수행한다.
    weights.assign_sub(lr * gradient)
```

### PyTorch Code

```python
import torch

# TensorFlow 예제와 같은 J(w)=(w-3)^2를 사용한다.
def compute_loss(weights):
    return (weights - 3.0).pow(2).mean()

# [1] torch.randn을 사용하여 평균 0, 표준편차 1인 정규분포에서 초기값을 무작위로 추출하고,
#     requires_grad=True를 설정하여 이 텐서가 학습 가능한 가중치(미분 대상)임을 명시한다
weights = torch.randn(1, requires_grad=True)
lr = 0.1

for step in range(20):
    # [2] 현재 가중치를 기반으로 비용 함수(Loss Function)인 compute_loss를 호출하여 손실 값을 계산한다
    #     파이토치는 텐서플로우와 달리 GradientTape 없이 일반 연산만으로 연산 그래프를 자동으로 기록한다
    loss = compute_loss(weights)
    
    # [3] loss.backward()를 호출하여 역전파(Backpropagation)를 수행한다
    # 이 연산을 통해 loss에 대한 weights의 편미분 값(Gradient)이 자동으로 계산되어 weights.grad에 저장된다
    loss.backward()
    
    # [4] torch.no_grad() 컨텍스트 내부에서 가중치를 업데이트한다
    # 경사하강법 공식에 따라 weights.grad 값을 이용해 가중치를 수동으로 갱신하며, 이 갱신 연산 자체는 미분 기록에서 제외한다
    with torch.no_grad():
        weights -= lr * weights.grad
        
        # [5] 파이토치는 미분 값이 누적되는 성질이 있으므로, 가중치를 업데이트한 후 
        #  다음 루프의 미분 계산을 위해 기존에 저장된 미분 값(.grad)을 반드시 0으로 초기화해 주어야 한다
        weights.grad.zero_()
```

### Computing Gradients: Backpropagation (역전파)

![back2](imagefile/back2.png)

$$\frac{\partial J(\boldsymbol{W})}{\partial w_2} = \frac{\partial J(\boldsymbol{W})}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial w_2}$$

- Chain Rule을 활용한 역전파 계산

- $\hat{y}$가 $w_2$에 직접 의존하는 단순한 경우이므로 두 미분 항의 곱으로 표현된다.

<br>

![back4](imagefile/back4.png)

$$\frac{\partial J(\boldsymbol{W})}{\partial w_1} = \frac{\partial J(\boldsymbol{W})}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial z_1} \cdot \frac{\partial z_1}{\partial w_1}$$

w2가 아닌 w1을 계산하고자 한다면?

- w1에 대한 J의 미분으로 바꾸어 본다

- 이전 식에 대비해서 한 번 더 미분해야 한다 (chain rule)

- 더 깊은 뉴럴 네트워크를 가지고 있다면, 계속 더 미분해주면 된다.

<br>

맨 끝에 있는 출력단에서 계산한 기울기들을 입력단이 있는 맨 앞까지 거꾸로 전파(Propagating)시킴으로써 이 과정을 반복할 수 있다

- 이렇게 하면 모든 각각의 가중치가 최종 손실에 어떤 영향을 미치는지 알아낼 수 있다

- 즉 gradient는 우리가 어떤 방향으로 움직였을 때 손실이 어떻게 변할지, 위로 올라갈지 아니면 아래로 내려갈지를 보여준다


<br>

실전에서는 이게 훨씬 더 복잡하다

실전에서의 뉴럴 네트워크의 Topology는 Highly non-convex(엄청난 비볼록) 형태를 띈다.

즉, 수많은 서로 다른 최솟값(Minimum)들이 존재한다.

어떤 초기화(Initialization)를 선택하는지, 어떤 최적화 방법과 Regularization을 적용하는지에 따라 학습 결과가 달라질 수 있다.
